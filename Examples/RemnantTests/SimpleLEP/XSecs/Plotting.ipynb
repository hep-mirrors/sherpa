```python
import glob, os

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.3})
def load_dat(path):
    xs, ys, errs = [], [], []
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith(("#", "//", "%")):
                continue
            tok = line.split()
            if not 2 <= len(tok) <= 3:
                continue
            try:
                vals = [float(t) for t in tok]
            except ValueError:
                continue
            xs.append(vals[0]); ys.append(vals[1])
            errs.append(vals[2] if len(vals) == 3 else np.nan)
    if not xs:
        raise ValueError(f"No numeric data found in {path}")
    x, y, e = map(np.array, (xs, ys, errs))
    order = np.argsort(x)
    return x[order], y[order], (None if np.all(np.isnan(e)) else e[order])
def plot_files(files, logx=False, logy=False, xlabel="k [GeV]", ylabel="y",
               title=None, xlim=None, ylim=None, separate=False, ncols=2,
               points=False, save=None):
    """Overlay files on one axis, or one subplot per file if separate=True."""
    files = list(files)
    n = len(files)
    if separate and n > 1:
        nrows = int(np.ceil(n / ncols))
        fig, axes = plt.subplots(nrows, ncols, figsize=(6*ncols, 4*nrows), squeeze=False)
        axes = axes.ravel()
        for extra in axes[n:]:
            extra.set_visible(False)
        axes = axes[:n]
    else:
        fig, ax = plt.subplots(figsize=(8, 5.5))
        axes = [ax] * n

    for ax, path in zip(axes, files):
        x, y, err = load_dat(path)
        label = os.path.splitext(os.path.basename(path))[0]
        style = "." if points else "-"
        if err is not None:
            ax.errorbar(x, y, yerr=err, fmt=style, lw=1.2, ms=3, label=label)
        else:
            ax.plot(x, y, style, lw=1.5, label=label)

    for ax in set(axes):
        if logx: ax.set_xscale("log")
        if logy: ax.set_yscale("log")
        if xlim: ax.set_xlim(*xlim)
        if ylim: ax.set_ylim(*ylim)
        ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.legend()
    if title:
        fig.suptitle(title)
    fig.tight_layout()
    if save:
        fig.savefig(save, dpi=200)
    plt.show()
    return fig

XSEC_DIR = "XSecs"
PATTERN  = "*ToD*.dat"   # matches pnToDGamma, nnToDPiMinus, pnToDPiZero, ppToDPiPlus

files = sorted(glob.glob(os.path.join(XSEC_DIR, PATTERN)))
for f in files:
    print(f)
```

* **k** is the paper's $k=|\vec p_1-\vec p_2|_{\rm COM}$ (Eq. (1)), i.e. *twice* the c.m. momentum of each nucleon.
* `expected_*`: every (anti)nucleon pair weighted with $P_i=\sigma_i/\sigma_0$, kinematics sampled 10 times — the $N_{\rm samp}\to\infty$ limit of the paper's weighting, Sec. III G.
  `formed_*`: deuterons actually formed in the events. Spectra are per event and per unit of the variable.
* `summary.dat`: per-event yields with errors.

```python
DATA_DIR  = "../DeuteronFormationFromXSecs"
VAL_DIR   = DATA_DIR + "/validation_invsigma0_18"
BABAR_DIR = DATA_DIR + "/BaBar_10.58GeV"

CHANNELS = ["pn_dgamma", "pn_dpi0", "pp_dpiplus", "nn_dpiminus",
            "pn_dpi0pi0", "pn_dpipluspiminus", "pp_dpipluspi0", "nn_dpiminuspi0"]
LABELS = {"pn_dgamma": r"$pn\to d\gamma$", "pn_dpi0": r"$pn\to d\pi^0$",
          "pp_dpiplus": r"$pp\to d\pi^+$", "nn_dpiminus": r"$nn\to d\pi^-$",
          "pn_dpi0pi0": r"$pn\to d\pi^0\pi^0$", "pn_dpipluspiminus": r"$pn\to d\pi^+\pi^-$",
          "pp_dpipluspi0": r"$pp\to d\pi^+\pi^0$", "nn_dpiminuspi0": r"$nn\to d\pi^-\pi^0$"}

def load_hist(name, directory=DATA_DIR):
    "Sherpa histogram file -> bin centres, values, errors, bin width (overflow row dropped)."
    x, y, e = load_dat(os.path.join(directory, name + ".dat"))
    x, y = x[:-1], y[:-1]
    e = np.zeros_like(y) if e is None else e[:-1]
    w = x[1] - x[0]
    return x + 0.5*w, y, e, w

def rebin(name, edges, directory=DATA_DIR, scale=1.):
    "Integrate a per-unit histogram over new bins; returns centres, per-unit values, errors."
    x, y, e, w = load_hist(name, directory)
    edges = np.asarray(edges); vals, errs = [], []
    for lo, hi in zip(edges[:-1], edges[1:]):
        m = (x > lo) & (x < hi)
        vals.append((y[m]*w).sum()/(hi-lo)); errs.append(np.sqrt(((e[m]*w)**2).sum())/(hi-lo))
    return 0.5*(edges[1:]+edges[:-1]), scale*np.array(vals), scale*np.array(errs)

def read_summary(directory=DATA_DIR):
    "summary.dat -> {key: (per-event mean, error)}, header lines."
    values, header = {}, []
    with open(os.path.join(directory, "summary.dat")) as f:
        for line in f:
            if line.startswith("#"):
                header.append(line[1:].strip()); continue
            tok = line.split()
            if len(tok) == 3: values[tok[0]] = (float(tok[1]), float(tok[2]))
    return values, header

def nevents(header):
    return float(header[0].split("events = ")[1])

summary, header = read_summary()
print("\n".join(header[:2]))
```

## 2. Available pairs at the Z pole — compare with Fig. 6 of arXiv:1504.07242 (p. 8)

Antinucleon pairs per event vs. $k$ (c.m.) and $|\vec p_1+\vec p_2|$ (lab), same axes, the three panels normalised together to a maximum of 1, as in Fig. 6 (left: $\bar p\bar n$, middle: $\bar p\bar p$, right: $\bar n\bar n$).
Statements in Sec. III E (p. 7–8) to check:
* "The distribution peaks for values of $k$ in the low GeV range and drops quickly for decreasing values of $k$."
* "The total number of available $\bar p\bar n$-pairs is roughly a factor two larger than the corresponding numbers of $\bar p\bar p$ and $\bar n\bar n$ pairs."

## 4. LEP: ALEPH and OPAL — compare with Sec. IV E and Tables VII/VIII (pp. 10, 13)

The paper has no LEP spectrum, only the two measurements:
* ALEPH: $(5.9\pm1.8\pm0.5)\times10^{-6}$ $\bar d$ per hadronic Z event for $0.62<p<1.03$ GeV, $|\cos\theta|<0.95$;
* OPAL: no candidate for $0.35<p<1.1$ GeV, with $s=\epsilon N_{\rm ev} n_{\bar d}$, $\epsilon=0.234$, $N_{\rm ev}=1.64\times10^6$ and $\chi^2_{\rm OPAL}=s$ (Eqs. (21), (22)).

The Sherpa result is to be compared with the best-fit $1/\sigma_0$ and $\chi^2_{\sigma_0}$ in Tables VII and VIII (p. 10), row "$\bar d$, LEP".
All yields scale exactly with $1/\sigma_0$, so the fit is analytic.

```python
inv_sigma0 = float(header[0].split("1/sigma_0 = ")[1].split()[0])
a, ae = summary["expected_dbar_ALEPH_window"]; a, ae = a/inv_sigma0, ae/inv_sigma0   # per unit 1/sigma_0
b, be = summary["expected_dbar_OPAL_window"];  b, be = b/inv_sigma0, be/inv_sigma0
A, sA = 5.9e-6, np.hypot(1.8e-6, 0.5e-6)
eps, Nev = 0.234, 1.64e6

chi2 = lambda x: (A - a*x)**2/sA**2 + eps*Nev*b*x
x_aleph = A/a
x_comb  = max(0., (A - eps*Nev*b*sA**2/(2*a))/a)      # d chi2/dx = 0
x_err   = sA/a                                         # Delta chi2 = 1 (the OPAL term is linear)

print(f"Sherpa at 1/sigma_0 = {inv_sigma0}/b: dbar in ALEPH window = {a*inv_sigma0:.3e} +- {ae*inv_sigma0:.1e},"
      f"  expected OPAL candidates s = {eps*Nev*b*inv_sigma0:.2f}")
print(f"1/sigma_0 from ALEPH alone      : {x_aleph:.2f} +- {x_err:.2f} /b")
lep_fit = (x_comb, x_err, chi2(x_comb))
print(f"1/sigma_0 from ALEPH+OPAL (paper): {x_comb:.2f} +- {x_err:.2f} /b,  chi2 = {lep_fit[2]:.2f}")
print( "-> compare with Tables VII and VIII of arXiv:1504.07242, row 'dbar, LEP'")
```

## 6. ALICE, pp at $\sqrt s$ = 0.9, 2.76, 7 TeV — compare with Fig. 7 of arXiv:1504.07242 (p. 10)

Same quantity and layout idea as Fig. 7: the invariant yield $\frac{1}{2\pi p_T N_{\rm ev}}\frac{d^2N}{dy\,dp_T}$ ($=E\,d^3N/dp^3$) at $|y|<0.5$ vs. $p_T$, one row per energy; 1/σ₀ fitted separately at each energy and for $d$ and $\bar d$, using only the experimental uncertainties (as the paper, Sec. IV).

* **Data**: the final ALICE results, HEPData ins1625294 (arXiv:1709.08522, Phys. Rev. C 97 (2018) 024615), Tables 1–3, stat ⊕ syst. They are normalised to *inelastic* events.
  Fig. 7 of the paper used the earlier, preliminary ALICE data of its Ref. [7], normalised to V0AND-triggered events (Eqs. (19), (20), Table IX), so the 1/σ₀ values are not one-to-one comparable with Tables VII/VIII.
* **Sherpa**: Amisic non-diffractive events (`Sherpa.Deuterons.ALICE_*TeV.yaml`), turned into per-inelastic-event yields with $f_{\rm ND}=1-\sigma_{\rm SD}/\sigma_{\rm inel}-\sigma_{\rm DD}/\sigma_{\rm inel}$ from the ALICE measurement, HEPData ins1181770 (arXiv:1208.4968), Tables 1–2 — assuming, as the paper does, that diffractive events produce no deuterons.

```python
import csv
ALICE_DATA = os.path.join(DATA_DIR, "data")
ENERGIES = [("0.9", 900., 1), ("2.76", 2760., 2), ("7", 7000., 3)]

def read_hepdata_blocks(path):
    "HEPData csv -> list of blocks, each a list of rows of floats (None for '-')."
    blocks, rows = [], []
    for line in open(path):
        line = line.strip()
        if not line or line.startswith("#"):
            if rows: blocks.append(rows); rows = []
            continue
        tok = next(csv.reader([line]))
        try: rows.append([None if v.strip() == "-" else float(v) for v in tok])
        except ValueError: continue          # column header line
    if rows: blocks.append(rows)
    return blocks

def alice_spectrum(table, block):
    "block 0: deuterons, 1: antideuterons -> pT edges, value, total error (stat+syst)."
    rows = [r for r in read_hepdata_blocks(os.path.join(ALICE_DATA, f"ALICE_pp_deuterons_HEPData_ins1625294_Table{table}.csv"))[block]
            if r[3] is not None]
    lo, hi, val = (np.array([r[i] for r in rows]) for i in (1, 2, 3))
    err = np.hypot(0.5*(np.array([r[4] for r in rows]) - np.array([r[5] for r in rows])),
                   0.5*(np.array([r[6] for r in rows]) - np.array([r[7] for r in rows])))
    return lo, hi, val, err

def f_nd(sqrts):
    out, err2 = 1., 0.
    for t in (1, 2):
        for r in read_hepdata_blocks(os.path.join(ALICE_DATA, f"ALICE_pp_diffraction_HEPData_ins1181770_Table{t}.csv"))[0]:
            if abs(r[0] - sqrts) < 1.:
                out -= r[3]; err2 += (0.5*(r[4] - r[5]))**2
    return out, np.sqrt(err2)

alice_fits = {}
fig, axes = plt.subplots(3, 2, figsize=(13, 13))
for row, (rs, sqrts, table) in enumerate(ENERGIES):
    adir = os.path.join(DATA_DIR, f"ALICE_{rs}TeV")
    asum, ahead = read_summary(adir)
    a_inv = float(ahead[0].split("1/sigma_0 = ")[1].split()[0])
    fnd, fnd_err = f_nd(sqrts)
    for col, (sp, block, lab) in enumerate([("d", 0, "d"), ("dbar", 1, r"\bar d")]):
        ax = axes[row, col]
        lo, hi, val, err = alice_spectrum(table, block)
        # each ALICE bin from its own edges (bins without a measurement are skipped)
        per_bin = [rebin(f"expected_{sp}_invyield_midy", [l, h], adir, scale=fnd) for l, h in zip(lo, hi)]
        mc, mce = np.array([b[1][0] for b in per_bin]), np.array([b[2][0] for b in per_bin])
        mc1, mce1 = mc/a_inv, mce/a_inv                      # per unit 1/sigma_0
        best = (val*mc1/err**2).sum()/(mc1**2/err**2).sum()
        best_err = 1./np.sqrt((mc1**2/err**2).sum())
        chi2_fit = (((val - best*mc1)/err)**2).sum()
        alice_fits[(rs, sp)] = (best, best_err, chi2_fit, len(val))

        ax.errorbar(0.5*(lo+hi), val, yerr=err, xerr=0.5*(hi-lo), fmt="o", ms=4, color="C0",
                    label="ALICE (HEPData ins1625294), stat $\\oplus$ syst")
        xs, ys, es = rebin(f"expected_{sp}_invyield_midy", np.arange(0., 4.01, 0.1), adir, scale=fnd)
        ok = ys > 0
        ax.step(xs[ok], best*ys[ok]/a_inv, where="mid", color="C3",
                label=f"Sherpa, best fit $1/\\sigma_0$ = {best:.2f}/b")
        ax.fill_between(xs[ok], best*(ys[ok]-es[ok])/a_inv, best*(ys[ok]+es[ok])/a_inv, step="mid", color="C3", alpha=0.25)
        ax.step(xs[ok], ys[ok], where="mid", color="C3", ls=":", lw=1, label=f"Sherpa, $1/\\sigma_0$ = {a_inv}/b")
        ax.set_yscale("log"); ax.set_xlim(0, 4); ax.set_ylim(1e-8, 1e-3)
        ax.set_xlabel(r"$p_T$ [GeV]"); ax.set_ylabel(r"$\frac{1}{2\pi p_T N_{\rm ev}}\frac{d^2N}{dy\,dp_T}$ [GeV$^{-2}$]")
        ax.set_title(f"${lab}$, pp at {rs} TeV, $|y|<0.5$  (cf. Fig. 7)"); ax.legend(fontsize=7)
    print(f"{rs:>5s} TeV: {nevents(ahead):.0f} ND events, f_ND = {fnd:.2f} +- {fnd_err:.2f}")
plt.tight_layout(); plt.show()

print(f"\n{'':10s} {'species':8s} {'1/sigma_0 [1/b]':>18s} {'chi2':>8s} {'points':>7s}")
for (rs, sp), (b, be, c2, n) in alice_fits.items():
    print(f"{rs+' TeV':10s} {sp:8s} {b:10.2f} +- {be:5.2f} {c2:8.1f} {n:7d}")
print("-> compare with Tables VII and VIII of arXiv:1504.07242, rows 'd, ALICE' and 'dbar, ALICE'"
      " (keeping in mind the different data set and normalisation, see above)")
```

## 8. Other formation models: HardDisk and Exponential coalescence

Same events (same run cards), but the formation probability per $pn$ pair is
* **HardDisk**: $P=1$ if $k<p_{\rm coal}$ (and $b<b_{\rm coal}$), else 0 — the classic coalescence model of Eq. (1) of arXiv:1504.07242;
* **Exponential**: $P=0.75\,e^{-b^2/C_b}\,e^{-k^2/C_p}$ with $C_{p,b}=p_{\rm coal}^2,\,b_{\rm coal}^2/(0.75\pi)$;

with $k=|\vec p_p-\vec p_n|$ in the pair frame and $b$ the impact parameter of the pair. The deuteron and a photon share the pair's four-momentum (the $pn\to d\gamma$ kinematics). Runs (from `SimpleLEP/`, e.g.)
`Sherpa -f Sherpa.Deuterons.yaml 'DEUTERON_FORMATION: {MODEL: HardDisk, P_COAL: 0.2, B_COAL: 0, OUTPUT: DeuteronFormationFromXSecs/coalescence/HardDisk_k/LEP}'`, with $p_{\rm coal}=0.2$ GeV and
* `*_k`: momentum condition only (`B_COAL: 0`), all setups;
* `*_kb`: also $b_{\rm coal}=3$ fm, LEP and ALICE 7 TeV.

As in Sec. IV A of the paper, the coalescence yield scales roughly as $p_0^3$, so one normalisation factor $s$ per data set corresponds to $p_0\simeq 0.2\,{\rm GeV}\times s^{1/3}$ — to be compared with the best-fit $p_0$ column of Tables VII/VIII of arXiv:1504.07242.

```python
# ALICE: dN/dy over the measured pT range, and dbar/d, vs sqrt(s)
def measured_dndy(table, block):
    lo, hi, val, err = alice_spectrum(table, block)
    xc = 0.5*(lo+hi)
    return (2*np.pi*xc*val*(hi-lo)).sum(), np.sqrt(((2*np.pi*xc*err*(hi-lo))**2).sum()), lo, hi
def sherpa_dndy(sp, adir, fnd, lo, hi):
    x, y, e, w = load_hist(f"expected_{sp}_invyield_midy", adir)
    m = np.zeros_like(x, dtype=bool)
    for l, h in zip(lo, hi): m |= (x > l) & (x < h)
    return fnd*(2*np.pi*x[m]*y[m]*w).sum(), fnd*np.sqrt(((2*np.pi*x[m]*e[m]*w)**2).sum())
rows = []
for rs, sqrts, table in ENERGIES:
    adir = os.path.join(DATA_DIR, f"ALICE_{rs}TeV"); fnd, _ = f_nd(sqrts)
    r = {"sqrts": sqrts/1000.}
    for sp, block in [("d", 0), ("dbar", 1)]:
        dv, de, lo, hi = measured_dndy(table, block)
        sv, se = sherpa_dndy(sp, adir, fnd, lo, hi)
        r[sp] = (dv, de, sv, se)
    rows.append(r)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.5))
for sp, c, lab in [("d", "C0", "d"), ("dbar", "C3", r"\bar d")]:
    xs = np.array([r["sqrts"] for r in rows])
    a1.errorbar(xs, [r[sp][0] for r in rows], yerr=[r[sp][1] for r in rows], fmt="o", color=c, label=rf"ALICE ${lab}$")
    a1.errorbar(xs*1.05, [r[sp][2] for r in rows], yerr=[r[sp][3] for r in rows], fmt="s", mfc="none", color=c,
                label=rf"Sherpa ${lab}$, $1/\sigma_0$ = 1.8/b")
a1.set_xscale("log"); a1.set_xlabel(r"$\sqrt{s}$ [TeV]"); a1.set_ylabel(r"$dN/dy$, $|y|<0.5$, measured $p_T$ range")
a1.legend(fontsize=8); a1.set_title("integrated yields (cf. Table 2 of arXiv:1709.08522)")
dr  = [r["dbar"][0]/r["d"][0] for r in rows]; dre = [q*np.hypot(r["dbar"][1]/r["dbar"][0], r["d"][1]/r["d"][0]) for q, r in zip(dr, rows)]
sr  = [r["dbar"][2]/r["d"][2] for r in rows]; sre = [q*np.hypot(r["dbar"][3]/r["dbar"][2], r["d"][3]/r["d"][2]) for q, r in zip(sr, rows)]
a2.errorbar(xs, dr, yerr=dre, fmt="o", color="k", label="ALICE")
a2.errorbar(xs*1.05, sr, yerr=sre, fmt="s", mfc="none", color="C3", label=r"Sherpa (independent of $\sigma_0$)")
a2.axhline(1, color="grey", lw=0.8); a2.set_xscale("log"); a2.set_ylim(0.5, 1.5)
a2.set_xlabel(r"$\sqrt{s}$ [TeV]"); a2.set_ylabel(r"$\bar d/d$"); a2.legend(fontsize=8); a2.set_title(r"$\bar d/d$ (cf. Fig. 6 of arXiv:1709.08522)")
plt.tight_layout(); plt.show()
for r in rows:
    print(f"{r['sqrts']:5.2f} TeV: d ALICE {r['d'][0]:.2e} +- {r['d'][1]:.1e}, Sherpa {r['d'][2]:.2e};  dbar ALICE {r['dbar'][0]:.2e}, Sherpa {r['dbar'][2]:.2e};"
          f"  dbar/d ALICE {r['dbar'][0]/r['d'][0]:.2f}, Sherpa {r['dbar'][2]/r['d'][2]:.2f}")
```

```python
# ALICE B2 vs pT/A (HEPData ins1625294, Tables 6-8): Sherpa at its best-fit 1/sigma_0 of Sec. 6
def sherpa_inv(name, adir, l, h):
    "Sherpa density averaged over [l, h]; for intervals narrower than a Sherpa bin, the bin containing it."
    x, y, e, w = load_hist(name, adir)
    m = (x > l) & (x < h)
    if not m.any(): m = np.abs(x - 0.5*(l+h)) <= 0.5*w
    return y[m].mean(), np.sqrt((e[m]**2).sum())/m.sum()
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5), sharey=True)
for ax, (rs, sqrts, table) in zip(axes, ENERGIES):
    adir = os.path.join(DATA_DIR, f"ALICE_{rs}TeV"); fnd, _ = f_nd(sqrts)
    a_inv = float(read_summary(adir)[1][0].split("1/sigma_0 = ")[1].split()[0])
    blocks = read_hepdata_blocks(os.path.join(ALICE_DATA, f"ALICE_pp_deuterons_HEPData_ins1625294_Table{5+table}.csv"))
    for (sp, psp, lab, c), blk in zip([("d", "p", "d", "C0"), ("dbar", "pbar", r"\bar d", "C3")], blocks):
        rows_ = [r for r in blk if r[3] is not None]
        lo, hi, val = (np.array([r[i] for r in rows_]) for i in (1, 2, 3))
        err = np.hypot(0.5*(np.array([r[4] for r in rows_])-np.array([r[5] for r in rows_])),
                       0.5*(np.array([r[6] for r in rows_])-np.array([r[7] for r in rows_])))
        scale = alice_fits[(rs, sp)][0]/a_inv
        b2, b2e = [], []
        for l, h in zip(lo, hi):
            dv, de = sherpa_inv(f"expected_{sp}_invyield_midy", adir, 2*l, 2*h)
            pv, pe = sherpa_inv(f"nucleon_{psp}_invyield_midy", adir, l, h)
            v = scale*dv/(fnd*pv**2) if pv > 0 else 0.
            b2.append(v); b2e.append(v*np.hypot(de/dv if dv else 0, 2*pe/pv if pv else 0))
        xc = 0.5*(lo+hi)
        ax.errorbar(xc, val, yerr=err, xerr=0.5*(hi-lo), fmt="o", color=c, label=rf"ALICE ${lab}$")
        ax.errorbar(xc+0.02, b2, yerr=b2e, fmt="s", mfc="none", color=c, label=rf"Sherpa ${lab}$")
    ax.set_title(f"pp {rs} TeV (cf. Fig. 8 of arXiv:1709.08522)"); ax.set_xlabel(r"$p_T/A$ [GeV]"); ax.legend(fontsize=8)
axes[0].set_ylabel(r"$B_2$ [GeV$^2$]"); axes[0].set_ylim(0, None)
plt.tight_layout(); plt.show()
print("Sherpa B2 at the best-fit 1/sigma_0 per energy and species (Sec. 6); B2 scales linearly with 1/sigma_0.")
```

```python
# ALEPH B2 (hep-ex/0604023, Eqs. (4), (5)) and the e+e- continuum limits
s_lep, h_lep = read_summary(DATA_DIR)
inv0 = float(h_lep[0].split("1/sigma_0 = ")[1].split()[0])
dOmega = 2*np.pi*2*0.95                                   # |cos(theta)| < 0.95
def inv_window(n, m, pmin, pmax):
    p0 = 0.5*(pmin+pmax); E = np.sqrt(p0**2+m**2)
    return n*E/(p0**2*dOmega*(pmax-pmin))
nd, nde = s_lep["expected_dbar_ALEPH_window"]; npb, npbe = s_lep["nucleon_pbar_ALEPH_halfwindow"]
inv_d, inv_p = inv_window(nd, 1.875613, 0.62, 1.03), inv_window(npb, 0.938272, 0.31, 0.515)
B2 = inv_d/inv_p**2; B2e = B2*np.hypot(nde/nd, 2*npbe/npb)
s_fit = lep_fit[0]/inv0                                   # ALEPH+OPAL best fit of Sec. 4
print(f"Sherpa LEP: dbar invariant yield {inv_d:.2e} GeV^-2 per event (ALEPH Eq. (4): (3.7 +- 1.1 +- 0.3)e-6 at its own rate)")
print(f"Sherpa LEP: antiproton (no weak decays) invariant yield at p/2: {inv_p:.3e} GeV^-2 per event")
print(f"Sherpa B2 = {B2*1e3:.2f} +- {B2e*1e3:.2f} e-3 GeV^2 at 1/sigma_0 = {inv0}/b;  {B2*s_fit*1e3:.2f} e-3 GeV^2 at the LEP best fit {lep_fit[0]:.2f}/b")
print( "ALEPH  B2 = 3.3 +- 1.0 +- 0.8 e-3 GeV^2  (hep-ex/0604023, Eq. (5))")
bsum, bhead = read_summary(BABAR_DIR)
b_inv = float(bhead[0].split("1/sigma_0 = ")[1].split()[0])
dtot, dtote = bsum["expected_dbar_total"]
t4 = read_hepdata_blocks(os.path.join(ALICE_DATA, "BaBar_continuum_ratio_HEPData_ins1286317_Table5.csv"))[0][0]
sig_had = 9.63e-3/t4[1]                                   # pb: BaBar sigma(dbar X) = 9.63 fb (HEPData Table 4) / ratio (Table 5)
print(f"\nContinuum e+e- -> q qbar at 10.58 GeV, Sherpa at 1/sigma_0 = {b_inv}/b:")
print(f"  dbar per hadronic event (all p)   : {dtot:.2e} +- {dtote:.1e}   ARGUS limit < 1.7e-5 (90% C.L., ~10 GeV, extrapolated in p)")
print(f"  sigma(e+e- -> dbar X) = {dtot*sig_had*1e3:.1f} fb  (sigma_had = {sig_had/1e3:.2f} nb from BaBar HEPData)   CLEO limit < 31 fb (90% C.L., 10.5 GeV);  BaBar measured 9.63 fb")
```

## 10. Multiplicity dependence at 7 TeV — ALICE, Phys. Lett. B 794 (2019) 50, [arXiv:1902.09290](https://arxiv.org/abs/1902.09290), HEPData ins1721729

Five runs of `Sherpa.Deuterons.ALICE_7TeV.yaml` (80k events each, `RANDOM_SEED` 1001–1005), output in `ALICE_7TeV_multiplicity/seed_*`, cross-section model.
* **Event classes** as in the paper (Sec. 2.2, Table 1): INEL>0 events (a charged particle in $|\eta|<1$) are ordered by the V0M estimator, here the number of charged final-state particles in $2.8\le\eta\le5.1$ and $-3.7\le\eta\le-1.7$, and split at the percentiles 0–4.7–9.5–19–38–100 % (classes I+II, III, IV+V, VI+VII, VIII+IX+X). Sherpa's percentiles are taken from its own (non-diffractive) INEL>0 events.
* **Observables**: $\langle dN_{ch}/d\eta\rangle_{|\eta|<0.5}$ per class (Table 1), deuteron spectra $\frac{1}{N_{ev}}\frac{d^2N}{dy\,dp_T}$ per class (HEPData Tables 1–5, paper Fig. 2), $2d/(p+\bar p)$ vs $\langle dN_{ch}/d\eta\rangle$ (Table 12, paper Fig. 8), and $B_2(p_T/A)$ per class (Tables 6–10, paper Fig. 4).
* One normalisation $1/\sigma_0$ is fitted to the deuteron spectra of **all classes together**; the multiplicity dependence is then a prediction, independent of $\sigma_0$.

```python
MULT_DIR = os.path.join(DATA_DIR, "ALICE_7TeV_multiplicity")
seeds = sorted(glob.glob(os.path.join(MULT_DIR, "seed_*")))
CLASSES = [("I+II", 0., 4.7, 17.47, 0.52), ("III", 4.7, 9.5, 13.50, 0.40), ("IV+V", 9.5, 19., 10.76, 0.30),
           ("VI+VII", 19., 38., 7.54, 0.23), ("VIII+IX+X", 38., 100., 3.30, 0.13)]   # arXiv:1902.09290, Table 1
SPECIES = ["d", "dbar", "p", "pbar"]
def load_rows(path):
    return {int(r[0]): np.array(r[1:]) for r in np.atleast_2d(np.loadtxt(path))}
ev, yl, y2 = {}, {sp: {} for sp in SPECIES}, {sp: {} for sp in SPECIES}
for sd in seeds:
    for k, v in load_rows(os.path.join(sd, "multiplicity_events.dat")).items(): ev[k] = ev.get(k, 0) + v
    for sp in SPECIES:
        for k, v in load_rows(os.path.join(sd, f"multiplicity_{sp}_invyield_midy.dat")).items(): yl[sp][k] = yl[sp].get(k, 0) + v
        for k, v in load_rows(os.path.join(sd, f"multiplicity_{sp}_invyield_midy_sq.dat")).items(): y2[sp][k] = y2[sp].get(k, 0) + v
nv0 = np.array(sorted(ev)); nevents = np.array([ev[k][0] for k in nv0])
# class weights per nV0 row: fraction of the row's events in each percentile class (highest nV0 = 0 %)
order = np.argsort(-nv0); cum_hi = np.cumsum(nevents[order])/nevents.sum()*100; cum_lo = cum_hi - nevents[order]/nevents.sum()*100
wts = np.zeros((len(CLASSES), len(nv0)))
for ic, (_, plo, phi, _, _) in enumerate(CLASSES):
    ov = np.clip(np.minimum(cum_hi, phi) - np.maximum(cum_lo, plo), 0, None)/np.maximum(cum_hi-cum_lo, 1e-12)
    wts[ic, order] = ov
cls = []
for ic, c in enumerate(CLASSES):
    w = wts[ic]; N = (w*nevents).sum()
    entry = {"name": c[0], "N": N, "nch": (w*np.array([ev[k][1] for k in nv0])).sum()/N,
             "alice_nch": c[3], "alice_nch_err": c[4]}
    for j, sp in enumerate(SPECIES):
        entry[sp+"_dndy"] = (w*np.array([ev[k][2+j] for k in nv0])).sum()/N
        entry[sp] = (w[:, None]*np.array([yl[sp][k] for k in nv0])).sum(axis=0)/N            # 1/(2 pi pT) d2N/dpTdy, 0.1 GeV bins
        entry[sp+"_err"] = np.sqrt((w[:, None]*np.array([y2[sp][k] for k in nv0])).sum(axis=0))/N
    cls.append(entry)
mult_inv0 = float(read_summary(seeds[0])[1][0].split("1/sigma_0 = ")[1].split()[0])
print(f"{len(seeds)} runs, {nevents.sum():.0f} INEL>0 events;  Sherpa <dNch/deta> all INEL>0: {sum(c['nch']*c['N'] for c in cls)/nevents.sum():.2f}  (ALICE 5.96 +- 0.23)")
print(f"{'class':10s} {'events':>8s} {'Sherpa <dNch/deta>':>19s} {'ALICE (Table 1)':>17s}")
for c in cls: print(f"{c['name']:10s} {c['N']:8.0f} {c['nch']:19.2f} {c['alice_nch']:11.2f} +- {c['alice_nch_err']:.2f}")
```

```python
# deuteron spectra per class (HEPData Tables 1-5): one 1/sigma_0 fitted to all classes together
pt_c = np.arange(0.05, 5.0, 0.1)
def class_dndpt(c, sp, l, h):
    m = (pt_c > l) & (pt_c < h)            # 1/N d2N/dy dpT averaged over [l, h]
    return (2*np.pi*pt_c[m]*c[sp][m]*0.1).sum()/(h-l), np.sqrt(((2*np.pi*pt_c[m]*c[sp+"_err"][m]*0.1)**2).sum())/(h-l)
data, model = [], []
for ic, c in enumerate(cls):
    lo, hi, val, err = (lambda b: (np.array([r[1] for r in b]), np.array([r[2] for r in b]), np.array([r[3] for r in b]),
        np.hypot(0.5*(np.array([r[4] for r in b])-np.array([r[5] for r in b])), 0.5*(np.array([r[6] for r in b])-np.array([r[7] for r in b])))))(
        [r for r in read_hepdata_blocks(os.path.join(ALICE_DATA, f"ALICE_pp7TeV_multiplicity_HEPData_ins1721729_Table{ic+1}.csv"))[0] if r[3] is not None])
    mc = np.array([class_dndpt(c, "d", l, h) for l, h in zip(lo, hi)])
    data.append((lo, hi, val, err)); model.append(mc)
allv = np.concatenate([d[2] for d in data]); alle = np.concatenate([d[3] for d in data]); allm = np.concatenate([m[:, 0] for m in model])
s_mult = (allv*allm/alle**2).sum()/(allm**2/alle**2).sum()
print(f"global fit to the deuteron spectra of all classes: 1/sigma_0 = {mult_inv0*s_mult:.2f}/b")
fig, axes = plt.subplots(1, 5, figsize=(20, 4.2), sharey=True)
for ax, c, (lo, hi, val, err), mc in zip(axes, cls, data, model):
    xc = 0.5*(lo+hi)
    chi2v = (((val - s_mult*mc[:, 0])/err)**2).sum()
    ax.errorbar(xc, val, yerr=err, xerr=0.5*(hi-lo), fmt="o", ms=3, color="k", label="ALICE")
    ax.errorbar(xc, s_mult*mc[:, 0], yerr=s_mult*mc[:, 1], xerr=0.5*(hi-lo), fmt="s", ms=3, color="C3", label="Sherpa")
    ax.set_yscale("log"); ax.set_xlabel(r"$p_T$ [GeV]")
    ax.set_title(f"class {c['name']}"); ax.text(0.05, 0.05, rf"$\chi^2$ = {chi2v:.0f} / {len(val)}", transform=ax.transAxes, fontsize=8)
axes[0].set_ylabel(r"$\frac{1}{N_{ev}}\frac{d^2N}{dy\,dp_T}$ [GeV$^{-1}$]"); axes[0].legend(fontsize=8)
fig.suptitle("deuterons per V0M class, pp 7 TeV (cf. Fig. 2 of arXiv:1902.09290)", y=1.02); plt.tight_layout(); plt.show()
```

```python
# 2d/(p+pbar) vs <dNch/deta> (HEPData Table 12, paper Fig. 8), Sherpa at the global 1/sigma_0
t12 = [r for r in read_hepdata_blocks(os.path.join(ALICE_DATA, "ALICE_pp7TeV_multiplicity_HEPData_ins1721729_Table12.csv"))[0] if r[3] is not None]
fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.5))
a1.errorbar([r[0] for r in t12], [r[3] for r in t12], xerr=[[r[0]-r[1] for r in t12], [r[2]-r[0] for r in t12]],
            yerr=[np.hypot(r[4], r[6]) for r in t12], fmt="o", color="k", label="ALICE (Table 12)")
sh = [2*s_mult*c["d_dndy"]/(c["p_dndy"]+c["pbar_dndy"]) for c in cls]
a1.plot([c["nch"] for c in cls], sh, "s-", color="C3", label=rf"Sherpa, $1/\sigma_0$ = {mult_inv0*s_mult:.2f}/b")
a1.set_xlabel(r"$\langle dN_{ch}/d\eta\rangle_{|\eta|<0.5}$"); a1.set_ylabel(r"$2d/(p+\bar p)$"); a1.set_ylim(0, None)
a1.legend(fontsize=8); a1.set_title("d/p vs multiplicity (cf. Fig. 8 of arXiv:1902.09290)")
# shape only: each normalised to its lowest-multiplicity value
a2.plot([r[0] for r in t12], np.array([r[3] for r in t12])/t12[0][3], "o", color="k", label="ALICE")
a2.plot([c["nch"] for c in cls][::-1], np.array(sh[::-1])/sh[-1], "s-", color="C3", label="Sherpa")
a2.set_xlabel(r"$\langle dN_{ch}/d\eta\rangle_{|\eta|<0.5}$"); a2.set_ylabel("relative to lowest class"); a2.legend(fontsize=8)
a2.set_title(r"trend only (independent of $\sigma_0$)")
plt.tight_layout(); plt.show()
```

```python
# B2(pT/A) per class (HEPData Tables 6-10, paper Fig. 4): (d+dbar)/2 over ((p+pbar)/2)^2, Sherpa at the global 1/sigma_0
fig, axes = plt.subplots(1, 5, figsize=(20, 4.2), sharey=True)
b2_fixed = []
for ic, (ax, c) in enumerate(zip(axes, cls)):
    rows_ = [r for r in read_hepdata_blocks(os.path.join(ALICE_DATA, f"ALICE_pp7TeV_multiplicity_HEPData_ins1721729_Table{6+ic}.csv"))[0] if r[3] is not None]
    lo, hi, val = (np.array([r[i] for r in rows_]) for i in (1, 2, 3)); err = np.array([np.hypot(r[4], r[6]) for r in rows_])
    b2, b2e = [], []
    for l, h in zip(lo, hi):
        md_ = (pt_c > 2*l) & (pt_c < 2*h); mp_ = (pt_c > l) & (pt_c < h)
        if not md_.any(): md_ = np.abs(pt_c - (l+h)) <= 0.05      # bin narrower than Sherpa's 0.1 GeV
        if not mp_.any(): mp_ = np.abs(pt_c - 0.5*(l+h)) <= 0.05
        dv = s_mult*0.5*(c["d"][md_]+c["dbar"][md_]).mean(); pv = 0.5*(c["p"][mp_]+c["pbar"][mp_]).mean()
        de = s_mult*0.5*np.sqrt((c["d_err"][md_]**2+c["dbar_err"][md_]**2).sum())/md_.sum()
        pe = 0.5*np.sqrt((c["p_err"][mp_]**2+c["pbar_err"][mp_]**2).sum())/mp_.sum()
        v = dv/pv**2 if pv > 0 else 0.; b2.append(v); b2e.append(v*np.hypot(de/dv if dv else 0, 2*pe/pv if pv else 0))
    xc = 0.5*(lo+hi)
    ax.errorbar(xc, val, yerr=err, xerr=0.5*(hi-lo), fmt="o", ms=3, color="k", label="ALICE")
    ax.errorbar(xc, b2, yerr=b2e, xerr=0.5*(hi-lo), fmt="s", ms=3, color="C3", label="Sherpa")
    ax.set_title(f"class {c['name']}"); ax.set_xlabel(r"$p_T/A$ [GeV]")
    sel = (lo >= 0.7-1e-9) & (hi <= 0.8+1e-9)
    if sel.any(): b2_fixed.append((c["nch"], c["alice_nch"], np.array(b2)[sel].mean(), val[sel].mean(), err[sel].mean()))
axes[0].set_ylabel(r"$B_2$ [GeV$^2$]"); axes[0].legend(fontsize=8); axes[0].set_ylim(0, None)
fig.suptitle("B2 per V0M class, pp 7 TeV (cf. Fig. 4 of arXiv:1902.09290)", y=1.02); plt.tight_layout(); plt.show()
if b2_fixed:
    fig, ax = plt.subplots(figsize=(7, 4.5))
    ax.errorbar([b[1] for b in b2_fixed], [b[3] for b in b2_fixed], yerr=[b[4] for b in b2_fixed], fmt="o", color="k", label="ALICE")
    ax.plot([b[0] for b in b2_fixed], [b[2] for b in b2_fixed], "s-", color="C3", label="Sherpa")
    ax.set_xlabel(r"$\langle dN_{ch}/d\eta\rangle_{|\eta|<0.5}$"); ax.set_ylabel(r"$B_2$ at $0.7<p_T/A<0.8$ GeV [GeV$^2$]")
    ax.set_title("B2 vs multiplicity (cf. Fig. 6 of arXiv:1902.09290)"); ax.legend(fontsize=8); ax.set_ylim(0, None)
    plt.tight_layout(); plt.show()
```
